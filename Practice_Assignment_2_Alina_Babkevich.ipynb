## Task 1 — Check whether a positive integer is a palindrome

Write a Python program to check if a given positive integer is a palindrome. Use mathematical operations only; do not convert the integer to a string.

```python
n = int(input("Enter a positive integer: "))

original = n
reverse = 0

while n > 0:
    digit = n % 10
    reverse = reverse * 10 + digit
    n = n // 10

if original == reverse:
    print("Palindrome")
else:
    print("Not a palindrome")
```

## Task 2 — Print integers in increasing row lengths

Take a positive integer `N` and print integers from 1 to `N`, with one integer on the first line, two on the second line, and so on.

```python
N = int(input("Enter N: "))

number = 1
row_length = 1

while number <= N:
    for i in range(row_length):
        if number <= N:
            print(number, end=" ")
            number += 1
    print()
    row_length += 1
```

## Task 4 — Find the least frequent character(s)

Count the frequency of each character in an input string and print the character or characters with the minimum frequency. This version ignores ordinary spaces.

```python
text = input("Enter a string: ")

frequency = {}

for char in text:
    if char != " ":
        if char in frequency:
            frequency[char] += 1
        else:
            frequency[char] = 1

if frequency:
    minimum = min(frequency.values())
    print("Minimum frequency:", minimum)
    print("Least frequent character(s):")

    for char, count in frequency.items():
        if count == minimum:
            print(repr(char))
else:
    print("No characters to count.")
```

## Task 5 — Convert infix notation to postfix notation

Convert a mathematical expression from infix format (for example, `A+B*C`) to postfix format (for example, `ABC*+`). Supports single-character letters/digits, parentheses, and `+`, `-`, `*`, `/`, `^` operators.

```python
def infix_to_postfix(expression):
    stack = []
    output = []

    precedence = {
        "+": 1,
        "-": 1,
        "*": 2,
        "/": 2,
        "^": 3
    }

    for char in expression:
        if char.isspace():
            continue

        if char.isalnum():
            output.append(char)

        elif char == "(":
            stack.append(char)

        elif char == ")":
            while stack and stack[-1] != "(":
                output.append(stack.pop())

            if not stack:
                raise ValueError("Mismatched parentheses")
            stack.pop()

        elif char in precedence:
            while (
                stack
                and stack[-1] != "("
                and (
                    precedence[stack[-1]] > precedence[char]
                    or (precedence[stack[-1]] == precedence[char] and char != "^")
                )
            ):
                output.append(stack.pop())

            stack.append(char)

        else:
            raise ValueError(f"Unsupported character: {char}")

    if "(" in stack:
        raise ValueError("Mismatched parentheses")

    while stack:
        output.append(stack.pop())

    return "".join(output)


expression = input("Enter an infix expression: ")
try:
    print("Postfix expression:", infix_to_postfix(expression))
except ValueError as error:
    print("Error:", error)
```

## Task 7 — Combine corresponding lines from two text files

Open two text files in read mode, combine corresponding lines, and write the results to a third file.

**Colab note:** You can create sample input files in the next cell, or upload your own files using the Colab Files panel.

```python

with open("file1.txt", "w") as file:
    file.write("Hello\nGood morning\n")

with open("file2.txt", "w") as file:
    file.write("World\nEveryone\n")

print("Sample files created: file1.txt and file2.txt")
```

```python
file1 = input("First filename [file1.txt]: ").strip() or "file1.txt"
file2 = input("Second filename [file2.txt]: ").strip() or "file2.txt"
output_file = input("Output filename [output.txt]: ").strip() or "output.txt"

with open(file1, "r", encoding="utf-8") as f1, \
     open(file2, "r", encoding="utf-8") as f2, \
     open(output_file, "w", encoding="utf-8") as out:

    for line1, line2 in zip(f1, f2):
        combined_line = line1.rstrip("\\n") + " " + line2.rstrip("\\n")
        out.write(combined_line + "\\n")

print(f"Combined lines written to {output_file}")
```

## Task 8 — Count words in a text file

Count words using the assignment's separators:
- one or more spaces;
- one or more newlines;
- a comma, period, or semicolon followed by a blank space;
- a hyphen preceded and followed by a blank space.

The regular expression below follows these separator rules literally.

```python
import re


sample_text = "Apples, oranges and pears.\\nFresh fruit - very healthy"
with open("sample.txt", "w", encoding="utf-8") as file:
    file.write(sample_text)

filename = input("Enter text filename [sample.txt]: ").strip() or "sample.txt"

with open(filename, "r", encoding="utf-8") as file:
    text = file.read()

pattern = r" +|\\n+|[,.;] +| +- +"
words = [word for word in re.split(pattern, text.strip()) if word]

print("Number of words:", len(words))
```
